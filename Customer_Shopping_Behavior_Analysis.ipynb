```python
import pandas as pd
from urllib.parse import quote_plus


print("Python is working!")
print(pd.__version__)
```

```python
df = pd.read_csv(r'E:\Data_Analytics\Projects\PowerBIAnudipProject\customer_shopping_behavior.csv')
df.head(10)
```

```python
df.info()
df.isnull().sum()
```

```python
# Fill null values in 'Review Rating' with the median of the respective 'Category'
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))

display(df.isnull().sum())
```

```python
df.columns = [col.lower().replace(' ', '_') for col in df.columns]
display(df.head())
```

```python
df.rename(columns={'purchase_amount_(usd)': 'purchase_amount_usd'}, inplace=True)
display(df.head())
df.columns
```

```python
# create a column age_group
labels = ["Young Adult", "Adult", "Middle_aged", "Senior"]
df['age_group']=pd.qcut(df['age'], q=4, labels=labels)

print("Value counts for 'age_group':")
display(df['age_group'].value_counts())

print("\nDataFrame head with new 'age_group' column:")
display(df.head())
```

```python
#create a column puechase_frequency_days
frequency_mapping = {
'Weekly': 7,
'Fortnightly': 14,
'Monthly': 30,
'Quarterly': 90,
'Annually':365,
'Bi-Weekly':14,
'Every 3 Months':90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
print("Value counts for 'purchase_frequency_days':")
display(df[['purchase_frequency_days','frequency_of_purchases']].head(50))

```
